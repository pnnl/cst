```python
import plotly.express as px
import pandas as pd
import json
import time
from pathlib import Path
import sys
import os
import numpy as np
```

```python
df = pd.read_csv("cst_scale1/timing_data.csv")
df = df.sort_values(by="name", ignore_index=True)
df["total_inputs"] = df.n_fedsxsubs + df.use_epts*df.n_fedsxsubs
df["log_time"] = df.time.apply(np.log)
df["log_feds"] = df.n_feds.apply(np.log)
df["log_subs"] = df.n_subs.apply(np.log)
df["log_fedsxsubs"] = df.n_fedsxsubs.apply(np.log)
df["log_inputs"] = df.total_inputs.apply(np.log)
```

```python
fig = px.scatter(
    df, 
    x="total_inputs", 
    y="time", 
    # facet_col="use_epts", 
    color="use_db", 
    hover_data=["name", "n_feds", "n_subs", "use_db", "use_pf", "time"], 
    symbol="n_feds",
    # facet_col="use_pf",
    log_x=True, log_y=True,
    # trendline="ols",
)
fig.show()
```

```python
fig = px.scatter(
    df, 
    x="log_inputs", 
    y="log_time", 
    # facet_col="use_epts", 
    color="use_db", 
    hover_data=["name", "n_feds", "n_subs", "use_epts", "use_db", "use_pf", "time"], 
    # symbol="n_feds",
    # log_x=True, log_y=True,
    trendline="ols",
)
fig.show()
results = px.get_trendline_results(fig)
# print(results)
```

```python
log_corr = df.corr().log_time
log_corr = log_corr.loc[["log_feds", "log_subs", "log_fedsxsubs", "log_inputs", "use_epts", "use_db", "use_pf"]]
fig = px.bar(log_corr).show()
```

```python
corr = df.corr().time
corr = corr.loc[["n_feds", "n_subs", "n_fedsxsubs", "total_inputs", "use_epts", "use_db", "use_pf"]]
fig = px.bar(corr).show()
```
